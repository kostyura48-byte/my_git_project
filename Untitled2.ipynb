```python
from transformers import pipeline


# Завантажуємо легкі моделі (працюють швидко навіть на слабких ПК)
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
generator = pipeline("text-generation", model="distilgpt2")


def ai_support_bot(user_review):
    # 1. Аналіз тональності
    analysis = classifier(user_review)[0]
    sentiment = analysis['label']

    # 2. Логіка вибору відповіді (ЗАВДАННЯ: змініть ці фрази)
    if sentiment == "POSITIVE":
        start_phrase = "We are thrilled you liked it! Regarding"
    else:
        start_phrase = "We apologize for the inconvenience. About"


    # 3. Генерація персоналізованого тексту
    # temperature=0.7 додає трохи випадковості, щоб відповіді не були однаковими
    response = generator(f"{start_phrase} {user_review[:40]}",
                         max_new_tokens=30,
                         temperature=0.7,
                         do_sample=True)

    return sentiment, response[0]['generated_text']


# Перевірка роботи
my_review = "The delivery was a bit late, but the quality is amazing!"
label, reply = ai_support_bot(my_review)


print(f"Аналіз: {label}")
print(f"Відповідь AI: {reply}")


```

```python
from transformers import pipeline

# Завантажуємо легкі моделі
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
generator = pipeline("text-generation", model="distilgpt2")

def ai_support_bot(user_review):
    # Переводимо в нижній регістр для зручності пошуку ключових слів
    review_lower = user_review.lower()

    # 1. Аналіз тональності
    analysis = classifier(user_review)[0]
    sentiment = analysis['label']

    # 2. Логіка вибору відповіді (Модифікована)

    # ЗАВДАННЯ: Додавання "Тригера на ціну"
    if "price" in review_lower or "expensive" in review_lower:
        start_phrase = "We always try to keep our prices fair. Let us explain..."

    # ЗАВДАННЯ: Зміна стилю (Positive/Negative)
    elif sentiment == "POSITIVE":
        start_phrase = "Wow, thanks for the good vibes! We're glad that"
    else:
        start_phrase = "Oh no, we're sorry to hear that. We'll look into"

    # 3. Генерація персоналізованого тексту
    # ЗАВДАННЯ: Експеримент з "хаосом" (temperature=1.5)
    response = generator(f"{start_phrase} {user_review[:40]}",
                         max_new_tokens=40,
                         temperature=1.5,
                         do_sample=True)

    return sentiment, response[0]['generated_text']

# Перевірка різних сценаріїв
reviews = [
    "This laptop is way too expensive for me!", # Тригер на ціну
    "The graphics card is super fast, I love it!", # Позитив
    "The cooling system is very loud and annoying." # Негатив
]

for my_review in reviews:
    label, reply = ai_support_bot(my_review)
    print(f"---")
    print(f"Відгук: {my_review}")
    print(f"Аналіз: {label}")
    print(f"Відповідь AI: {reply}")
```
